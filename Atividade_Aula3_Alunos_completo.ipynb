```python
import pandas as pd
```

```python
path = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
formacao_bruto = pd.read_excel(path, header=None)
```

```python
formacao_bruto
```

A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar	

Exemplo 1

```python
# : = todas as linhas  |  0:4 = colunas 0, 1, 2 e 3 (as 4 primeiras)
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

**Exemplo 2**

Agora as colunas de Ciência, Tecnologia, Engenharias e Matemática (colunas 4, 5 e 6), mantendo a coluna 0 (nome da região).

```python
# [0] = nome da região  |  [4,5,6] = total, homens e mulheres da área CTEM
formacao_ctem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
```

```python
formacao_ctem.columns = [
    "uf_regiao",
    "total_ctem",
    "homens_ctem",
    "mulheres_ctem"
]
```

```python
formacao_ctem
```

```python
formacao_ctem["percentual_mulheres_ctem"] = ((formacao_ctem['mulheres_ctem'] / formacao_ctem['total_ctem']) * 100).round(2)
formacao_ctem
```

**Exemplo 3**

Agora as colunas de Educação, Serviços pessoais, Saúde e Bem-estar (colunas 7, 8 e 9), mantendo também a coluna 0.

```python
# [0] = nome da região  |  [7,8,9] = total, homens e mulheres da área Educação/Saúde
formacao_edu_saude = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
```

```python
formacao_edu_saude.columns = [
    "uf_regiao",
    "total_edu_saude",
    "homens_edu_saude",
    "mulheres_edu_saude"
]
```

```python
formacao_edu_saude
```

```python
formacao_edu_saude["percentual_mulheres_edu_saude"] = ((formacao_edu_saude['mulheres_edu_saude'] / formacao_edu_saude['total_edu_saude']) * 100).round(2)
formacao_edu_saude
```

**Comparando as 3 tabelas**

No total do curso superior mulheres já são maioria. Vamos comparar o % de mulheres em cada uma das duas áreas específicas com o Brasil.

```python
brasil_total = formacao_total.loc[formacao_total['uf_regiao'] == 'Brasil'].iloc[0]
brasil_ctem = formacao_ctem.loc[formacao_ctem['uf_regiao'] == 'Brasil'].iloc[0]
brasil_edu = formacao_edu_saude.loc[formacao_edu_saude['uf_regiao'] == 'Brasil'].iloc[0]

print("Brasil - % de mulheres com curso superior concluído")
print(f"  no total (todas as áreas):                 {brasil_total.percentual_mulheres}%")
print(f"  em Ciência, Tecnologia, Eng. e Matemática:  {brasil_ctem.percentual_mulheres_ctem}%")
print(f"  em Educação, Serviços pessoais e Saúde:     {brasil_edu.percentual_mulheres_edu_saude}%")
```
